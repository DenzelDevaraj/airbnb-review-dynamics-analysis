```python
# Import the PuLP library
import pulp

# Filter for failing students (G3 < 10)
failing_students_df = pres_dfSam[pres_dfSam['G3'] < 10].copy()

# Define the problem
prob = pulp.LpProblem("Tutoring_Resource_Allocation", pulp.LpMaximize)

# Decision variables: xi = 1 if student i is selected for tutoring, 0 otherwise
# Use the index of the failing students DataFrame as the identifier
student_indices = failing_students_df.index.tolist()
x = pulp.LpVariable.dicts("student", student_indices, cat='Binary')

# Objective function: Maximize the number of failing students selected for tutoring
# Since we've already filtered for failing students, each selected student contributes 1 to the objective
prob += pulp.lpSum(x[i] for i in student_indices), "Total Failing Students Tutored"

# Constraint: Tutoring Hours Budget
# ci = a + absences_i / 10, where a is a base cost (let's assume a=1 for simplicity)
# B = 80 (Total available tutoring hours)
base_cost = 1
failing_students_df['tutoring_cost'] = base_cost + failing_students_df['absences'] / 10
budget = 80

prob += pulp.lpSum(failing_students_df.loc[i, 'tutoring_cost'] * x[i] for i in student_indices) <= budget, "Budget Constraint"

# Solve the problem
prob.solve()

# Print the results
print("Status:", pulp.LpStatus[prob.status])

print("\nStudents selected for tutoring:")
tutored_students = [i for i in student_indices if x[i].varValue == 1]
print(tutored_students)

print("\nNumber of failing students tutored:", pulp.value(prob.objective))

print("\nTotal tutoring hours used:", sum(failing_students_df.loc[i, 'tutoring_cost'] for i in tutored_students))
```
